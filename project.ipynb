```python
class observation:
    def __init__(self, target, observer,  filter_name, exposure):
        self.target = target
        self.observer = observer
        self.filter = filter_name
        self.exposure = exposure
    def summary_info(self):
        return f" target: {self.target}\n filter name: {self.filter}\n exposure: {self.exposure}\n observer name: {self.observer}"

    def exposure_min(self):
        return self.exposure / 60
    def save_to_db(self, cursor, conn):
        cursor.execute("""
        INSERT INTO observation
        (target, observer, filter_name, exposure)
        VALUES(?,?,?,?)""",
                      (self.target,
                      self.observer,
                      self.filter,
                      self.exposure))
        conn.commit()
class Photometric_observation(observation):
    def __init__(self, target, observer,  filter_name, exposure, wavelength):
        super().__init__(target, observer,  filter_name, exposure)
        self.wavelength = wavelength
    def summary_info(self):
        return super().summary_info() + f"\n wavelength : {self.wavelength}"
```

```python
from pathlib import Path

code = '''import sqlite3


def save_to_db(observations):
    connection = sqlite3.connect("observation.db")
    cursor = connection.cursor()

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS observations (
            target TEXT,
            observer TEXT,
            filter_name TEXT,
            exposure REAL
        )
    """)

    for obs in observations:
        cursor.execute(
            "INSERT INTO observations VALUES (?, ?, ?, ?)",
            (obs.target, obs.observer, obs.filter, obs.exposure)
        )

    connection.commit()
    connection.close()
'''

Path("database.py.txt").write_text(code, encoding="utf-8")

print("Database code saved!")
```

```python
from observation import Observation
from database import save_to_db
from visualization import show_graph


observations = []


def add_observation():
    target = input("Target name: ")
    observer = input("Observer name: ")
    filter_name = input("Filter name: ")
    exposure = float(input("Exposure time in seconds: "))

    obs = Observation(target, observer, filter_name, exposure)
    observations.append(obs)
    print("Observation added successfully!")


def show_observations():
    for obs in observations:
        print(obs.summary_info())


def search_observation():
    target = input("Enter target name: ")

    for obs in observations:
        if obs.target.lower() == target.lower():
            print(obs.summary_info())
            return

    print("Observation not found.")

def average_exposure():
    if not observations:
        print("No observations available.")
        return
    total = sum(obs.exposure for obs in observations)
    average = total / len(observations)
    print(f"Average exposure: {average} seconds")


def max_exposure():
    if not observations:
        print("No observations available.")
        return
    maximum = max(obs.exposure for obs in observations)
    print(f"Maximum exposure: {maximum} seconds")


def min_exposure():
    if not observations:
        print("No observations available.")
        return
    minimum = min(obs.exposure for obs in observations)
    print(f"Minimum exposure: {minimum} seconds")


def total_exposure():
    if not observations:
        print("No observations available.")
        return
    total = sum(obs.exposure for obs in observations)
    print(f"Total exposure: {total} seconds")
    
def main():
    while True:
        print("\n1. Add observation")
        print("2. Show observations")
        print("3. Search observation")
        print("4. Save to database")
        print("5. Show graph")
        print("6. Average exposure")
        print("7. Maximum exposure")
        print("8. Minimum exposure")
        print("9. Total exposure")
        print("10. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            add_observation()
        elif choice == "2":
            show_observations()
        elif choice == "3":
            search_observation()
        elif choice == "4":
            save_to_db(observations)
            print("Observations saved.")
        elif choice == "5":
            show_graph(observations)
        elif choice == "6":
            average_exposure()
        elif choice == "7":
            max_exposure()
        elif choice == "8":
            min_exposure()
        elif choice == "9":
            total_exposure()
        elif choice == "10":
            break
        else:
            print("Invalid choice.")


if __name__ == "__main__":
    main()
```
